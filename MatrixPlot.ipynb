```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
```

```python
flight = sns.load_dataset("flights")
flight
```

```python
pivot_flights = flight.pivot(index = "month", columns = "year", values = "passengers")
```

```python
# Heatmap

sns.heatmap(
    pivot_flights,
    cmap = "coolwarm",
    annot = True,
    fmt = "d"
)

plt.title("Passengers heatmap")
plt.tight_layout()
```
