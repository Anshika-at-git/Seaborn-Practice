```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
tips = sns.load_dataset("tips")
tips.head()
```

```python
# Bar Chart

sns.barplot(
    data = tips,
    x = "day",
    y = "tip",
    hue = "sex"
)
```

```python
# Box Plot

sns.boxplot(
    data = tips,
    x = "day",
    y = "tip",
)
```
